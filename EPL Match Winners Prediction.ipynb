```python
def rolling_averages(group, cols, new_cols):
    #sort group by date to be in ascending order for last 3 matches
    group = group.sort_values("Date")
    #compute rolling averages and closed = left takes current week out
    rolling_stats = group[cols].rolling(3, closed='left').mean()
    group[new_cols] =  rolling_stats
    group = group.dropna(subset=new_cols)
    return group
```

```python
cols = ['GF','GA','Sh','SoT','Dist','FK','PK','PKatt']
#create new column
new_cols = [f"{c}_rolling" for c in cols]
```

```python
new_cols
```

```python
rolling_averages(group, cols, new_cols)
```

```python
#apply to all matches which applies 1 function to each team
matches_rolling = matches.groupby("Team").apply(lambda x: rolling_averages(x, cols, new_cols))
```

```python
matches_rolling
```

```python
matches_rolling = matches_rolling.droplevel('Team')
```

```python
#Unique values in index
matches_rolling.index = range(matches_rolling.shape[0])
```

```python
def predictions(data, predictors):
    train_data =  data[data['Date'] < '2022-01-01']
    test_data = data[data['Date'] > '2022-01-01']
    rf = RF(n_estimators = 50, min_samples_split = 10, random_state = 1)
    rf.fit(train_data[predictors], train_data["Target"])
    preds = rf.predict(test_data[predictors])
    combined = pd.DataFrame(dict(actual=test_data['Target'], prediction = preds))
    prec_score = precision_score(test_data['Target'], preds)
    return combined, prec_score
```

```python
#GF_rolling	GA_rolling	Sh_rolling	SoT_rolling	Dist_rolling	FK_rolling	PK_rolling	PKatt_rolling
combined, precision = predictions(matches_rolling, predictors + new_cols)
```
